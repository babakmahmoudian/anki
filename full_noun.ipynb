```python
import pandas as pd
```

```python
df = pd.read_csv('./nouns.csv', sep='\t', header=None)

# ['deutsch', 'english', 'aussprache', 'genus', 'plural', 'biespeal', 'deutsch_audio', 'plural_audio', 'full_noun']

df
```

```python
def merg_func(row):
    genus=row[3]
    deutch=row[0]
    return f"{genus} {deutch}" if genus else deutch

df[8] = df[[0,3]].fillna('').apply(merg_func, axis='columns')
```
